```python
# ============================================================
# 16. Chronological Train-Test Split
# ============================================================

# Sort by year
model_df = merged_df.sort_values("year").reset_index(drop=True)

# Training: 1997-2016
# Testing: 2017-2020
train_df = model_df[model_df["year"] <= 2016].copy()
test_df = model_df[model_df["year"] >= 2017].copy()

features = [
    "crop",
    "state",
    "year",
    "season",
    "area",
    "fertilizer",
    "pesticide"
]

target = "yield"

X_train = train_df[features].copy()
y_train = train_df[target].copy()

X_test = test_df[features].copy()
y_test = test_df[target].copy()

print("=" * 60)
print("CHRONOLOGICAL TRAIN-TEST SPLIT")
print("=" * 60)

print("Training records:", len(X_train))
print("Testing records :", len(X_test))

print("\nTraining year range:")
print(X_train["year"].min(), "to", X_train["year"].max())

print("\nTesting year range:")
print(X_test["year"].min(), "to", X_test["year"].max())

print("\nTraining target statistics:")
print(y_train.describe())

print("\nTesting target statistics:")
print(y_test.describe())
```

```python
# ============================================================
# 17. Verify No Temporal Leakage
# ============================================================

print("Maximum training year:", X_train["year"].max())
print("Minimum testing year :", X_test["year"].min())

assert X_train["year"].max() < X_test["year"].min()

print("\n✅ No temporal overlap between training and testing data.")
```

```python
# ============================================================
# 18. Preprocessing Pipeline
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

categorical_features = [
    "crop",
    "state",
    "season"
]

numeric_features = [
    "year",
    "area",
    "fertilizer",
    "pesticide"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "numeric",
            SimpleImputer(strategy="median"),
            numeric_features
        )
    ]
)

print("Categorical features:", categorical_features)
print("Numeric features:", numeric_features)

print("\n✅ Preprocessing pipeline created successfully.")
```

```python
# ============================================================
# 19. Baseline Random Forest - Raw Yield
# ============================================================

from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

baseline_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                n_jobs=-1,
                max_features="sqrt"
            )
        )
    ]
)

print("Training baseline model...")

baseline_model.fit(X_train, y_train)

print("✅ Baseline model trained.")
```

```python
# ============================================================
# 20. Baseline Model Evaluation
# ============================================================

y_pred_baseline = baseline_model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred_baseline)
rmse = np.sqrt(mean_squared_error(y_test, y_pred_baseline))
r2 = r2_score(y_test, y_pred_baseline)

print("=" * 60)
print("BASELINE RANDOM FOREST RESULTS")
print("=" * 60)

print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")
```

```python
# ============================================================
# 21. Random Forest with Log-Transformed Target
# ============================================================

from sklearn.compose import TransformedTargetRegressor

log_rf = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                n_jobs=-1,
                max_features="sqrt"
            )
        )
    ]
)

# Transform target only during training.
# Predictions are automatically converted back to original yield.
log_model = TransformedTargetRegressor(
    regressor=log_rf,
    func=np.log1p,
    inverse_func=np.expm1
)

print("Training log-target Random Forest...")

log_model.fit(X_train, y_train)

print("✅ Log-target model trained.")
```

```python
# ============================================================
# 22. Log-Target Model Evaluation
# ============================================================

y_pred_log = log_model.predict(X_test)

# Prevent tiny negative numerical predictions
y_pred_log = np.maximum(y_pred_log, 0)

mae_log = mean_absolute_error(y_test, y_pred_log)
rmse_log = np.sqrt(mean_squared_error(y_test, y_pred_log))
r2_log = r2_score(y_test, y_pred_log)

print("=" * 60)
print("LOG-TARGET RANDOM FOREST RESULTS")
print("=" * 60)

print(f"MAE  : {mae_log:.4f}")
print(f"RMSE : {rmse_log:.4f}")
print(f"R²   : {r2_log:.4f}")
```

```python
# ============================================================
# 23. Model Comparison
# ============================================================

comparison = pd.DataFrame({
    "Model": [
        "Random Forest - Raw Yield",
        "Random Forest - Log Yield"
    ],
    "MAE": [
        mae,
        mae_log
    ],
    "RMSE": [
        rmse,
        rmse_log
    ],
    "R2": [
        r2,
        r2_log
    ]
})

print(comparison)
```

```python
# ============================================================
# 24. Generate Final Baseline Predictions
# ============================================================

y_pred = baseline_model.predict(X_test)

# Prevent impossible negative yield predictions
y_pred = np.maximum(y_pred, 0)

print("Predictions generated successfully.")
print("Number of predictions:", len(y_pred))
print("Number of test records:", len(y_test))
```

```python
# ============================================================
# 25. Detailed Test-Set Evaluation
# ============================================================

results = pd.DataFrame({
    "actual": y_test.values,
    "predicted": y_pred
})

results["error"] = (
    results["actual"] - results["predicted"]
)

results["absolute_error"] = (
    results["error"].abs()
)

results["percentage_error"] = np.where(
    results["actual"] != 0,
    (results["absolute_error"] / results["actual"]) * 100,
    np.nan
)

print("=" * 70)
print("DETAILED RANDOM FOREST TEST RESULTS")
print("=" * 70)

mae = mean_absolute_error(
    results["actual"],
    results["predicted"]
)

rmse = np.sqrt(
    mean_squared_error(
        results["actual"],
        results["predicted"]
    )
)

r2 = r2_score(
    results["actual"],
    results["predicted"]
)

print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")

print("\nError statistics:")
print(results["absolute_error"].describe())
```

```python
# ============================================================
# 26. Worst Prediction Analysis
# ============================================================

test_results = test_df[
    [
        "crop",
        "state",
        "year",
        "season",
        "area",
        "fertilizer",
        "pesticide",
        "yield"
    ]
].copy()

test_results["predicted_yield"] = y_pred

test_results["absolute_error"] = (
    test_results["yield"] -
    test_results["predicted_yield"]
).abs()

worst_predictions = (
    test_results
    .sort_values("absolute_error", ascending=False)
    .head(20)
)

worst_predictions
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    median_absolute_error
)

# ============================================================
# TIME-BASED RANDOM FOREST - RESIDUAL ANALYSIS
# ============================================================

error_analysis = X_test.copy()

error_analysis["actual_yield"] = np.asarray(y_test)
error_analysis["predicted_yield"] = np.asarray(y_pred_time)

# Residual: actual - predicted
error_analysis["error"] = (
    error_analysis["actual_yield"]
    - error_analysis["predicted_yield"]
)

error_analysis["absolute_error"] = (
    error_analysis["error"].abs()
)

# Avoid division by zero
epsilon = 1e-6

error_analysis["relative_error_pct"] = (
    error_analysis["absolute_error"]
    / np.maximum(error_analysis["actual_yield"], epsilon)
) * 100

print("=" * 70)
print("TIME-BASED RANDOM FOREST - ERROR ANALYSIS")
print("=" * 70)

print(f"Records analysed : {len(error_analysis)}")
print(f"MAE              : {mean_absolute_error(y_test, y_pred_time):.4f}")
print(f"RMSE             : {np.sqrt(mean_squared_error(y_test, y_pred_time)):.4f}")
print(f"R²               : {r2_score(y_test, y_pred_time):.4f}")
print(f"Median AE        : {median_absolute_error(y_test, y_pred_time):.4f}")

print("\n" + "=" * 70)
print("ERROR DISTRIBUTION")
print("=" * 70)

print(
    error_analysis["absolute_error"].describe(
        percentiles=[0.50, 0.75, 0.90, 0.95, 0.99]
    )
)

print("\n" + "=" * 70)
print("BIAS / RESIDUAL STATISTICS")
print("=" * 70)

print(f"Mean residual     : {error_analysis['error'].mean():.4f}")
print(f"Median residual   : {error_analysis['error'].median():.4f}")
print(f"Std residual      : {error_analysis['error'].std():.4f}")

print("\nPositive residual = model UNDERPREDICTED")
print("Negative residual = model OVERPREDICTED")
```

```python
import numpy as np
import pandas as pd

print("=" * 70)
print("CREATING YEAR-BASED TIME-SERIES CROSS-VALIDATION")
print("=" * 70)

# ------------------------------------------------------------
# Unique years in training data
# ------------------------------------------------------------
train_years = np.sort(X_train["year"].unique())

print("Training years:")
print(train_years)

# ------------------------------------------------------------
# Define non-overlapping chronological folds
# ------------------------------------------------------------
time_folds = [
    {
        "train_years": train_years[train_years <= 2004],
        "val_years": train_years[(train_years >= 2005) & (train_years <= 2008)]
    },
    {
        "train_years": train_years[train_years <= 2008],
        "val_years": train_years[(train_years >= 2009) & (train_years <= 2012)]
    },
    {
        "train_years": train_years[train_years <= 2012],
        "val_years": train_years[(train_years >= 2013) & (train_years <= 2016)]
    }
]

# ------------------------------------------------------------
# Verify folds
# ------------------------------------------------------------
for i, fold in enumerate(time_folds, 1):

    train_year_set = set(fold["train_years"])
    val_year_set = set(fold["val_years"])

    overlap = train_year_set.intersection(val_year_set)

    print("\nFold", i)
    print("-" * 50)

    print("Training years :", list(fold["train_years"]))
    print("Validation years:", list(fold["val_years"]))

    print("Training records:",
          X_train["year"].isin(fold["train_years"]).sum())

    print("Validation records:",
          X_train["year"].isin(fold["val_years"]).sum())

    print("Year overlap:", overlap)

    if len(overlap) == 0:
        print("✅ No temporal overlap")
    else:
        print("❌ TEMPORAL OVERLAP")
```

```python
cv_splits = []

for fold in time_folds:

    train_idx = X_train.index[
        X_train["year"].isin(fold["train_years"])
    ]

    val_idx = X_train.index[
        X_train["year"].isin(fold["val_years"])
    ]

    # Convert dataframe index labels to positional indices
    train_positions = X_train.index.get_indexer(train_idx)
    val_positions = X_train.index.get_indexer(val_idx)

    cv_splits.append((train_positions, val_positions))

print("=" * 70)
print("CUSTOM TIME CV CREATED")
print("=" * 70)

for i, (train_idx, val_idx) in enumerate(cv_splits, 1):

    print(f"\nFold {i}")
    print("Training records :", len(train_idx))
    print("Validation records:", len(val_idx))

    train_years_fold = X_train.iloc[train_idx]["year"].unique()
    val_years_fold = X_train.iloc[val_idx]["year"].unique()

    print("Training years :", sorted(train_years_fold))
    print("Validation years:", sorted(val_years_fold))

    overlap = set(train_years_fold).intersection(set(val_years_fold))

    print("Overlap:", overlap)

    assert len(overlap) == 0

print("\n✅ CUSTOM CV PASSED TEMPORAL LEAKAGE CHECK")
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

print("=" * 70)
print("STARTING RANDOM FOREST HYPERPARAMETER SEARCH")
print("=" * 70)

# ------------------------------------------------------------
# Random Forest
# ------------------------------------------------------------
rf = RandomForestRegressor(
    random_state=42,
    n_jobs=-1
)

# ------------------------------------------------------------
# Pipeline
# ------------------------------------------------------------
rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", rf)
])

# ------------------------------------------------------------
# Parameter distributions
# ------------------------------------------------------------
param_distributions = {
    "model__n_estimators": [200, 300, 500, 700],
    
    "model__max_depth": [
        None,
        10,
        15,
        20,
        25,
        30
    ],
    
    "model__min_samples_split": [
        2,
        5,
        10
    ],
    
    "model__min_samples_leaf": [
        1,
        2,
        4
    ],
    
    "model__max_features": [
        0.5,
        0.7,
        1.0
    ]
}

# ------------------------------------------------------------
# Search
# ------------------------------------------------------------
rf_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_distributions,
    n_iter=15,
    scoring="neg_mean_absolute_error",
    cv=cv_splits,
    random_state=42,
    n_jobs=-1,
    verbose=2,
    return_train_score=True
)

rf_search.fit(X_train, y_train)

print("\n" + "=" * 70)
print("HYPERPARAMETER SEARCH COMPLETE")
print("=" * 70)

print("\nBest parameters:")
print(rf_search.best_params_)

print("\nBest CV MAE:")
print(-rf_search.best_score_)
```

```python
# ------------------------------------------------------------
# Best tuned model
# ------------------------------------------------------------
best_rf = rf_search.best_estimator_

# ------------------------------------------------------------
# Test predictions
# ------------------------------------------------------------
y_pred_tuned = best_rf.predict(X_test)

# ------------------------------------------------------------
# Metrics
# ------------------------------------------------------------
mae_tuned = mean_absolute_error(y_test, y_pred_tuned)

rmse_tuned = np.sqrt(
    mean_squared_error(y_test, y_pred_tuned)
)

r2_tuned = r2_score(
    y_test,
    y_pred_tuned
)

median_ae_tuned = np.median(
    np.abs(y_test.values - y_pred_tuned)
)

print("=" * 70)
print("TUNED RANDOM FOREST — FINAL TEST RESULTS")
print("=" * 70)

print(f"MAE         : {mae_tuned:.4f}")
print(f"RMSE        : {rmse_tuned:.4f}")
print(f"R²          : {r2_tuned:.4f}")
print(f"Median AE   : {median_ae_tuned:.4f}")
```
