```python
def predict_soil_health(
    nitrogen,
    phosphorus,
    potassium,
    soil_moisture,
    humidity,
    soil_ph
):
    
    # High stress conditions
    if (
        nitrogen < 20
        or phosphorus < 15
        or potassium < 15
        or soil_moisture < 15
        or soil_ph < 5.5
        or soil_ph > 7.5
    ):
        return "High Stress"

    # Moderate stress conditions
    elif (
        nitrogen < 30
        or phosphorus < 25
        or potassium < 25
        or soil_moisture < 25
        or soil_ph < 6.0
        or soil_ph > 7.0
    ):
        return "Moderate Stress"

    # Otherwise
    return "Healthy"
```

```python
np.random.seed(42)

base_values = {
    "Nitrogen_Level": 30,
    "Phosphorus_Level": 30,
    "Potassium_Level": 35,
    "Soil_Moisture": 30,
    "Humidity": 55,
    "Soil_pH": 6.5
}

noise_results = []

for i in range(100):
    
    noisy = {
        key: value + np.random.normal(0, value * 0.02)
        for key, value in base_values.items()
    }
    
    prediction = predict_soil_health(
        noisy["Nitrogen_Level"],
        noisy["Phosphorus_Level"],
        noisy["Potassium_Level"],
        noisy["Soil_Moisture"],
        noisy["Humidity"],
        noisy["Soil_pH"]
    )
    
    noise_results.append(prediction)

pd.Series(noise_results).value_counts()
```

```python
baseline_prediction = predict_soil_health(
    30,
    30,
    35,
    30,
    55,
    6.5
)

print("Baseline prediction:", baseline_prediction)
```

```python
stability = (
    pd.Series(noise_results) == baseline_prediction
).mean()

print(f"Prediction stability: {stability * 100:.2f}%")
```

```python
for sensor, base_value in base_values.items():

    predictions = []

    for _ in range(100):
        noisy_value = base_value + np.random.normal(
            0,
            base_value * 0.02
        )

        test_values = base_values.copy()
        test_values[sensor] = noisy_value

        prediction = predict_soil_health(
            test_values["Nitrogen_Level"],
            test_values["Phosphorus_Level"],
            test_values["Potassium_Level"],
            test_values["Soil_Moisture"],
            test_values["Humidity"],
            test_values["Soil_pH"]
        )

        predictions.append(prediction)

    print(f"\n{sensor}")
    print(pd.Series(predictions).value_counts())
```

```python
baseline_prediction = predict_soil_health(
    30, 30, 35, 30, 55, 6.5
)

print("Baseline prediction:", baseline_prediction)
```

```python
np.random.seed(42)

for sensor, base_value in base_values.items():

    predictions = []

    for _ in range(100):
        
        test_values = base_values.copy()

        # Add 2% Gaussian sensor noise to ONLY this sensor
        noisy_value = base_value + np.random.normal(
            0,
            base_value * 0.02
        )

        test_values[sensor] = noisy_value

        prediction = predict_soil_health(
            test_values["Nitrogen_Level"],
            test_values["Phosphorus_Level"],
            test_values["Potassium_Level"],
            test_values["Soil_Moisture"],
            test_values["Humidity"],
            test_values["Soil_pH"]
        )

        predictions.append(prediction)

    prediction_counts = pd.Series(predictions).value_counts()
    stability = prediction_counts.get(
        baseline_prediction, 0
    ) / len(predictions)

    print("=" * 50)
    print(sensor)
    print(prediction_counts)
    print(f"Stability: {stability * 100:.2f}%")
```

```python
nitrogen_test_values = np.arange(20, 41, 0.5)

nitrogen_results = []

for nitrogen in nitrogen_test_values:

    prediction = predict_soil_health(
        nitrogen,
        30,   # Phosphorus
        35,   # Potassium
        30,   # Soil Moisture
        55,   # Humidity
        6.5   # Soil pH
    )

    nitrogen_results.append({
        "Nitrogen": nitrogen,
        "Prediction": prediction
    })

nitrogen_boundary_df = pd.DataFrame(nitrogen_results)

display(nitrogen_boundary_df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(
    nitrogen_boundary_df["Nitrogen"],
    nitrogen_boundary_df["Prediction"].map({
        "Healthy": 1,
        "Moderate Stress": 0,
        "High Stress": -1
    }),
    marker="o"
)

plt.yticks(
    [-1, 0, 1],
    ["High Stress", "Moderate Stress", "Healthy"]
)

plt.xlabel("Nitrogen Level")
plt.ylabel("Soil Health Status")
plt.title("Nitrogen Sensitivity Analysis")

plt.grid(True)
plt.show()
```

```python
sensor_ranges = {
    "Nitrogen_Level": np.arange(20, 41, 0.5),
    "Phosphorus_Level": np.arange(20, 41, 0.5),
    "Potassium_Level": np.arange(20, 41, 0.5),
    "Soil_Moisture": np.arange(15, 41, 0.5),
    "Humidity": np.arange(40, 71, 0.5),
    "Soil_pH": np.arange(5.5, 7.6, 0.1)
}

base_values = {
    "Nitrogen_Level": 30,
    "Phosphorus_Level": 30,
    "Potassium_Level": 35,
    "Soil_Moisture": 30,
    "Humidity": 55,
    "Soil_pH": 6.5
}

for sensor, values in sensor_ranges.items():

    print("\n" + "=" * 60)
    print(sensor)

    previous_prediction = None

    for value in values:

        test_values = base_values.copy()
        test_values[sensor] = value

        prediction = predict_soil_health(
            test_values["Nitrogen_Level"],
            test_values["Phosphorus_Level"],
            test_values["Potassium_Level"],
            test_values["Soil_Moisture"],
            test_values["Humidity"],
            test_values["Soil_pH"]
        )

        if prediction != previous_prediction:
            print(
                f"{sensor} = {value:.2f}"
                f" → {prediction}"
            )

        previous_prediction = prediction
```

## Conclusion

The prototype soil-health rule engine was developed using
hardware-relevant soil and environmental parameters.

The rules were evaluated using boundary and extreme-value
testing to verify consistent behavior.

The current thresholds are dataset-derived prototype rules
and should not be interpreted as universal agronomic thresholds.
Field validation is required before real-world agricultural
deployment.
