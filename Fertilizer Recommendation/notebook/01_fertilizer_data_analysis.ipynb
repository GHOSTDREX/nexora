```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.metrics import accuracy_score, classification_report

X_diag = fertilizer_df[
    [
        "Nitrogen_Level",
        "Phosphorus_Level",
        "Potassium_Level"
    ]
]

y_diag = fertilizer_df["Recommended_Fertilizer"]

X_train_diag, X_test_diag, y_train_diag, y_test_diag = train_test_split(
    X_diag,
    y_diag,
    test_size=0.20,
    random_state=42,
    stratify=y_diag
)

diag_tree = DecisionTreeClassifier(
    max_depth=4,
    random_state=42,
    class_weight="balanced"
)

diag_tree.fit(X_train_diag, y_train_diag)

diag_pred = diag_tree.predict(X_test_diag)

print("NPK-only diagnostic accuracy:")
print(round(accuracy_score(y_test_diag, diag_pred), 4))

print("\nClassification Report:")
print(classification_report(
    y_test_diag,
    diag_pred,
    zero_division=0
))

print("\nLearned Rules:")
print(
    export_text(
        diag_tree,
        feature_names=list(X_diag.columns)
    )
)
```

```python
diagnostic_features = [
    "Soil_pH",
    "Soil_Moisture",
    "Organic_Carbon",
    "Electrical_Conductivity",
    "Nitrogen_Level",
    "Phosphorus_Level",
    "Potassium_Level",
    "Temperature",
    "Humidity",
    "Rainfall",
    "Fertilizer_Used_Last_Season",
    "Yield_Last_Season"
]

full_range_analysis = (
    fertilizer_df
    .groupby("Recommended_Fertilizer")[diagnostic_features]
    .agg(["min", "max", "mean"])
    .round(2)
)

display(full_range_analysis)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report
)

X_diag2 = fertilizer_df[
    [
        "Nitrogen_Level",
        "Phosphorus_Level",
        "Potassium_Level",
        "Soil_pH",
        "Electrical_Conductivity",
        "Organic_Carbon",
        "Soil_Moisture"
    ]
]

y_diag2 = fertilizer_df["Recommended_Fertilizer"]

X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X_diag2,
    y_diag2,
    test_size=0.20,
    random_state=42,
    stratify=y_diag2
)

diag_tree2 = DecisionTreeClassifier(
    max_depth=7,
    class_weight="balanced",
    random_state=42
)

diag_tree2.fit(X_train2, y_train2)

pred2 = diag_tree2.predict(X_test2)

print("Accuracy:")
print(round(accuracy_score(y_test2, pred2), 4))

print("\nBalanced Accuracy:")
print(round(balanced_accuracy_score(y_test2, pred2), 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test2,
        pred2,
        zero_division=0
    )
)

print("\nLearned Rules:")
print(
    export_text(
        diag_tree2,
        feature_names=list(X_diag2.columns),
        max_depth=7
    )
)
```

```python
feature_importance = pd.DataFrame({
    "Feature": X_diag2.columns,
    "Importance": diag_tree2.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
)

display(feature_importance)
```

## Dataset Analysis Conclusion

- Original dataset size: 10,000 records.
- Client-specific dataset contains 2,857 records:
  - Rice: 1,429
  - Sugarcane: 1,428
- Target variable: Recommended_Fertilizer.
- Seven fertilizer classes are available:
  Urea, DAP, MOP, Compost, NPK, SSP, Zinc Sulphate.
- The target is imbalanced, especially for SSP.
- Diagnostic Decision Tree analysis showed that the fertilizer labels are
  strongly determined by Soil pH, Nitrogen, Phosphorus and Potassium.
- Soil Moisture and Organic Carbon provided almost no predictive contribution
  in the diagnostic tree.
- Rice and Sugarcane show very similar fertilizer-class distributions.
- The dataset therefore appears to follow structured/synthetic fertilizer
  recommendation rules rather than representing fully validated crop-specific
  field observations.
- The dataset will be used for prototype ML experimentation and decision-support
  development only.
- Real agronomic validation is required before operational fertilizer advice.
